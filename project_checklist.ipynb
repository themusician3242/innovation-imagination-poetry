```python
# Building the Programmatic Checklist:
c = {
     'Figma Wireframe' : '⌛',
     'Website Header & Footer': '⌛',
     'Website Hero' : '⌛',
     'Website Sections' : '⌛',
     'Website Icon' : '⌛',
     'Other Graphical Design Assets': '⌛',
     'Website Transition': '⌛',
     'Data Backend': '⌛',
     'Rive Interactive UI': '⌛',
     'Testing & Deployment' : '⌛',
}

class Checklist: 
     def __init__(self, checklist):
          self.checklist = checklist

     def __call__(self):
          items = "\n".join(f"[{status}] {name}" for name, status in self.checklist.items())
          print(f"""=========== CHECKLIST ===========
{items}
=================================""")

     def add(self, newItem): 
          self.checklist[newItem] = '⌛'
     
     def check(self):
          for key, value in self.checklist.items():
               if value != '✅':
                    self.checklist[key] = '✅'
                    break

     def clear(self):
          for key, value in self.checklist.items():
               self.checklist[key] = '⌛'

     def check_at_index(self, index):
          newIndex = index - 1
          for key, value in self.checklist.items():
               if newIndex == list(self.checklist.keys()).index(key):
                    self.checklist[key] = '✅'

checklist = Checklist(c)    
```

```python
checklist()
```
